### 概要  
3つの整数a,b,cと文字列sを入力、a+b+cの計算結果とsを並べて出力する。

```python
a = int(input())
b, c = map(int, input().split())
s = input()

print("{} {}".format(a+b+c, s))
```

map(関数, リストやタプル(複数の値を順番に並べて1つにまとめたデータ構造))

## ABC086A - Product  
- **問題リンク:** [ABC086A - Product](https://atcoder.jp/contests/abs/tasks/abc086_a)  
  
### 概要  
正の整数a,bの積が偶数か奇数か判定する。

```python
a, b = map(int,input().split())
product = a * b
if product % 2 == 0:
  print("Even")
else:
  print("Odd")
```

### 📝振り返り  

- 自力で解くことができた。ロジックは自分で考えてpythonでのif文の書き方だけ調べた  
- 一応三項演算子というものが存在していてA if 条件 else Bと書くと一行で納められる  
print("Odd" if (a * b) % 2 != 0 else "Even")

## ABC081B - Shift only  
- **問題リンク:** [ABC081B - Shift only](https://atcoder.jp/contests/abs/tasks/abc081_b)  
  
### 概要  
N個の正の整数がありすべて偶数の時2で割ったものに置き換える。これを何回出来るかを出力する。


```python
n = int(input())
numbers = map(int, input().split())
count = 0
s = 10000

for number in numbers:
  count = 0
  while number % 2 == 0:
    count += 1
    number = number / 2
  if count < s:
    s = count

print(s)
```

### 📝振り返り  

- すごく時間はかかったが自力で解くことができてうれしかった  
- whileの書き方は調べた。ifをどこに入れるかsの初期設定をどうするかを工夫することでうまく動いた。
- 別解  



```python
def count_twos(n): #nにaがそれぞれ入る  
    count = 0  
    while n % 2 == 0:  
        count += 1  
        n //= 2    
    return count #割れた回数countを返す  

N = int(input())  
A = map(int, input().split())  

print(min(count_twos(a) for a in A)) 
```

返したcountがリストに入りmin()でその中の最小値を出力する。  
Pythonの/は浮動小数点数floatを返すため繰り返すと精度誤差のリスクがある。整数の割り算には//を使う。
