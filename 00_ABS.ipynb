## ABC081B - Shift only  
- **問題リンク:** [ABC081B - Shift only](https://atcoder.jp/contests/abs/tasks/abc081_b)  
  
### 概要  
N個の正の整数がありすべて偶数の時2で割ったものに置き換える。これを何回出来るかを出力する。


```python
n = int(input())
numbers = map(int, input().split())
count = 0
s = 10000

for number in numbers:
  count = 0
  while number % 2 == 0:
    count += 1
    number = number / 2
  if count < s:
    s = count

print(s)
```

### 📝振り返り  

- すごく時間はかかったが自力で解くことができてうれしかった  
- whileの書き方は調べた。ifをどこに入れるかsの初期設定をどうするかを工夫することでうまく動いた。
- 別解  



```python
def count_twos(n): #nにaがそれぞれ入る  
    count = 0  
    while n % 2 == 0:  
        count += 1  
        n //= 2    
    return count #割れた回数countを返す  

N = int(input())  
A = map(int, input().split())  

print(min(count_twos(a) for a in A)) 
```

返したcountがリストに入りmin()でその中の最小値を出力する。  
Pythonの/は浮動小数点数floatを返すため繰り返すと精度誤差のリスクがある。整数の割り算には//を使う。

## ABC083B - Some Sums  
- **問題リンク:** [ABC083B - Some Sums](https://atcoder.jp/contests/abs/tasks/abc083_b)  
  
### 概要  
1以上N以下の整数のうち、10進数での各桁の和がA以上B以下であるものの総和を出力する。

```python
N, A, B = map(int, input().split())
SUM = 0

for n in range(1, N+1, 1):
  sum = 0
  Q = n
  r = 0
  while Q != 0:
    r = Q % 10
    Q //= 10
    sum += r
  if sum >= A and sum <= B:
    SUM += n

print(SUM)
```

### 📝振り返り  

- 今回も自力で解くことができた  
- かつがandまたはがor否定がnot優先順位はnot>and>orの順  
- rとQの計算の順番、sum,Q,rの初期化のタイミング最初whileの中でnを使っていて計算が終わらなかった  
- sum()という組み込み関数が存在するのでtotalやdigit_sumを使った方がよかった  
- PythonではA <= total <= Bと直感的に書ける  
- 数字を一度文字列strに変換するとsum(map(int, str(n)))で一瞬で解ける

## ABC088B - Card Game for Two  
- **問題リンク:** [ABC088B - Card Game for Two](https://atcoder.jp/contests/abs/tasks/abc088_b)  
  
### 概要  
N枚のカードがあり数字がそれぞれ書かれている。交互にとっていって、最初に取り始めた人が何点多く取れるか。それぞれ最適な戦略をとっている。

```python
N = int(input())
numbers = list(map(int, input().split()))
A = 0
B = 0
count = 1

while numbers:
  if count % 2 != 0:
    A += max(numbers)
    numbers.remove(max(numbers))
    count += 1
  else:
    B += max(numbers)
    numbers.remove(max(numbers))
    count +=1

print(A-B)
```

### 📝振り返り  

- 自力でロジックは考えられた。  
- input().split()だけでは文字列整数にするのにmapを使いそれをリスト化することで要素を消したりすることができる。文字列だと数値を比較したりできない。  
- mapは言われたらその都度変換して1個ずつ出すみたいなイメージ消したりできない  
- removeで配列の要素を消す、list、maxについて調べた。pythonのリストは空であればFalse,入っていればTrueを返す。not リストやlen(リスト)==0で空となる。
- カードを大きい順にソートしてやることもできる

reverse=Trueは大きい順(逆順)デフォルトは小さい順になっている。デフォルトの場合は書く必要ない。
- スライスsum(numbers[::2])  
[::2]で２個進み(1個飛ばし)

## ABC085B - Kagami Mochi  
- **問題リンク:** [ABC085B - Kagami Mochi](https://atcoder.jp/contests/abs/tasks/abc085_b)  
  
### 概要  
N枚のもちを直径が大きい順に何段並べられるかを出力する。同じ大きさのものは並べられない。

```python
N = int(input())
D = []
m = 1000
count = 0

for i in range(N):
  D.append(int(input()))

for d in sorted(D, reverse=True):
  if d < m:
    m = d
    count += 1

print(count)
```

### 📝振り返り  

- 自力で短時間で解くことができた。  
- iterable(イテラブル)エラーが最初出ていたのをリストに入れることで修正した。for文ではイテラブルでない数値を指定できない。つまりリストや文字列でないものは返せない。

## ABC085C - Otoshidama  
- **問題リンク:** [ABC085C - Otoshidama](https://atcoder.jp/contests/abs/tasks/abc085_c)  
  
### 概要  
お札がN枚合計がY円。10000,5000,1000札で合計が作れたらその一例を、なかったら-1 -1 -1を出力する。

```python
N, Y = map(int, input().split())
i = -1
j = -1
k = -1

for x in range(N, -1, -1):
  if i != -1 and j != -1 and k != -1:
    break
  y = 0
  while y <= N-x:
    z = N-x-y
    if (10000*x) + (5000*y) + (1000*z) == Y:
      i = x
      j = y
      k = z
      break
    y += 1       

print("{} {} {}".format(i, j, k))
```
