## PracticeA - Welcome to AtCoder  
- **問題リンク:** [PracticeA - Welcome to AtCoder](https://atcoder.jp/contests/abs/tasks/practice_1)  
  
### 概要  
3つの整数a,b,cと文字列sを入力、a+b+cの計算結果とsを並べて出力する。

```python
a = int(input())
b, c = map(int, input().split())
s = input()

print("{} {}".format(a+b+c, s))
```

### 📝振り返り  

- 自力で解けなかったので解説を参照  
- 入力についてinput()を使う。整数の場合int()としたり、文字列の場合はinput()のみでいい  
- map(int, input().split())でスペース区切りの整数入力ができる  
input().split()この部分で入力したものをスペース区切りにしている  
map(関数, リストやタプル(複数の値を順番に並べて1つにまとめたデータ構造))  
- 出力の仕方print("{} {}".format(a+b+c, s))

## ABC086A - Product  
- **問題リンク:** [ABC086A - Product](https://atcoder.jp/contests/abs/tasks/abc086_a)  
  
### 概要  
正の整数a,bの積が偶数か奇数か判定する。

```python
a, b = map(int,input().split())
product = a * b
if product % 2 == 0:
  print("Even")
else:
  print("Odd")
```

### 📝振り返り  

- 自力で解くことができた。ロジックは自分で考えてpythonでのif文の書き方だけ調べた  
- 一応三項演算子というものが存在していてA if 条件 else Bと書くと一行で納められる  
print("Odd" if (a * b) % 2 != 0 else "Even")
