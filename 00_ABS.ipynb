### 概要  
3つの整数a,b,cと文字列sを入力、a+b+cの計算結果とsを並べて出力する。

```python
a = int(input())
b, c = map(int, input().split())
s = input()

print("{} {}".format(a+b+c, s))
```

### 📝振り返り  

- 自力で解けなかったので解説を参照  
- 入力についてinput()を使う。整数の場合int()としたり、文字列の場合はinput()のみでいい  
- map(int, input().split())でスペース区切りの整数入力ができる  
input().split()この部分で入力したものをスペース区切りにしている  
map(関数, リストやタプル(複数の値を順番に並べて1つにまとめたデータ構造))  
- 出力の仕方print("{} {}".format(a+b+c, s))
