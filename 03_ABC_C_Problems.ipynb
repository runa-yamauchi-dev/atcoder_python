```python
N = int(input())
X = 0
X_lis = []
ans = 0

for x in sorted(map(int, input().split())):
  X += x
  X_lis.append(x)

P = round(X/N)

for x in X_lis:
  ans += (x-P)*(x-P)

print(ans)
```

- round()で四捨五入 
- より安全に四捨五入する場合
