## C - Rally 【Easy 100 #002】
🟢 **Status: 自力AC**(20分)

```python
N = int(input())
X = 0
X_lis = []
ans = 0

for x in sorted(map(int, input().split())):
  X += x
  X_lis.append(x)

P = round(X/N)

for x in X_lis:
  ans += (x-P)*(x-P)

print(ans)
```

- <span style="color: red;">round()</span>で四捨五入 
- より安全に四捨五入する場合  
P = <span style="color: red;">int(sum(X_lis) / N + 0.5)</span>  
0.5加えて切り捨て
- sorted今回いらない。思考途中で入れたままだった。
- 2乗<span style="color: red;">(x-P)**2</span>でいい

## C - String Too Long 【2026/8/3 ADT EASY2】
🔴 **Status: 解説AC**(60分)  
🔴 **Status: 解説AC**(40分)

```python
N = int(input())
cl = []
C = []
L = []
S = []
ans = []
count = 0

for i in range(N):
  cl = list(map(str, input().split()))
  C.append(cl[0])
  L.append(cl[1])

while sum(map(int, L)) <= 100 and count < N-1:
  for c in C:
    for i in range(int(L[count])):
      ans.append(c)
    count += 1

if sum(map(int, L)) <= 100:
  result = ''.join(ans)
  print(result)
else:
  print("Too Long")
```

- ループの回数が多すぎたことが時間をオーバーする理由だった。  
- <span style="color: red;">result = ''.join(ans)</span>の書き方は調べた。リストから文字列にしている。','とかにすると,で区切れる。
- 解説の解答

```python
N = int(input())
P = []
for i in range(N):
    c, l = input().split()
    P.append([c, int(l)])
if sum(map(lambda x: x[1], P)) > 100:
    print("Too Long")
else:
    S = ""
    for c, l in P:
        S += c * l
    print(S)
```

<span style="color: red;">P.append([c, int(l)])</span>で方が違くても同時に入れている。  
<span style="color: red;">lambda x(引数)</span>['a', 3]のようなペアを取りだし<span style="color: red;">: x[1](処理内容)</span>そのペアの1番目の要素。これにより数字だけを取り出して計算している。  
<span style="color: red;">S = ""</span>で空っぽの文字列を作れる。  
<span style="color: red;">for c, l in P:</span>でぺアを自動で2つの変数にばらして受け取れる。['a', 3]がそのままではなく。  
文字列に数字を掛けえるとその回数分リピートされる。  <span style="color: red;">S +=</span>で新しい文字を後ろにくっつける。  
この解法から新しいことをたくさん学ぶことができた。

```python
N = int(input())
ans = ""

for _ in range(N):
  s, k = input().split()
  k = int(k)
  if len(ans) + k > 100:
    ans = "Too Long"
    break
  ans += s * k

print(ans)
```

- リストにする必要なかった

## C - Traveling Salesman around Lake 【Easy 100 #012】
🟢 **Status: 自力AC**(40分)

```python
K, N = map(int, input().split())
befor_a = 0
count = 0
max_a = 0
total = 0

for a in map(int, input().split()):
  if count != 0:
    if max_a < (a - before_a):
      max_a = a - before_a
    total += a - before_a
  before_a = a
  count += 1

if max_a < K - total:
  print(K - K + total)
else:
  print(K- max_a)
```

## C - Go to School 【Easy 100 #016】
🔴 **Status: 解説AC**(26分)

```python
N = int(input())
A = list(map(int, input().split()))
ans = ""
count = 0

for n in range(1, N+1):
  a = str(A.index(n) + 1)
  ans += a
  count += 1
  if count != N:
    ans += " "

print(ans)
```

- 時間がオーバーする理由が全く分からなかった
- A.index(n)これがN回ループしているのと一緒だった
- <span style="color: red;">list.index(x)</span>で配列の中のxの位置がわかる。ただxがその配列にない場合エラーになる  
- なのでそれを使わない別解

```python
N = int(input())
A = list(map(int, input().split()))

# サイズ N のリストをあらかじめ作っておく
ans = [0] * N

for i in range(N):
  ans[A[i] - 1] = i + 1

# リストの要素をスペース区切りで一括出力
print(*ans)
```

A[i]の要素から1引けば順番になる  
要素は2なら1,3なら2,1なら0  
入る数字はi+1  
<span style="color: red;">*ans</span>でスペース区切りにできる

## C - Alchemist 【Easy 100 #017】
🟠 **Status: 実装もたつきAC**(75分) 

```python
import itertools

N = int(input())
V = list(map(int, input().split()))

while len(V) != 1:
  min = 10000000000000
  p =[]
  q = []
  for pair in itertools.combinations(V, 2):
    if sum(pair)/2 < min:
      p, q = map(float, pair)
      min = sum(pair)/2
  V.remove(p)
  V.remove(q)
  V.append(min)

for v in V:
  print(v)
```

- ロジックを考えるのにすごく時間がかかった。平均が最小になる組み合わせを選んでいけばいいことに確信が持てなかった。例外もあるのではと疑ってしまった。  
- <span style="color: red;">import itertools</span>で<span style="color: red;">combinations(p,r)</span>が使えるようになる。p組み合わせを見たいもの,r何個で組み合わせたいか2だとペアの組み合わせだけ。  
- p, q = map(float, pair)いつものくせでintにしていたせいでREが出てしまったが自分で解決することができた  
- 価値の小さい具材ほど先に合成して何度も 2 で割り、価値の大きい具材は後半に合成して減額を抑えるというのが得策であっていた。深く考えすぎていた。なので極端にこのような書き方もできる  

```python
N = int(input())
V = list(map(int, input().split()))

# 小さい順に並べ替える
V.sort()

# 一番小さい要素から順に合成していく
ans = V[0]
for i in range(1, N):
    ans = (ans + V[i]) / 2

print(ans)
```
