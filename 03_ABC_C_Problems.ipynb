## C - Rally 【Easy 100 #002】
🟢 **Status: 自力AC**(20分)

```python
N = int(input())
X = 0
X_lis = []
ans = 0

for x in sorted(map(int, input().split())):
  X += x
  X_lis.append(x)

P = round(X/N)

for x in X_lis:
  ans += (x-P)*(x-P)

print(ans)
```

- round()で四捨五入 
- より安全に四捨五入する場合  
P = int(sum(X_lis) / N + 0.5)
0.5加えて切り捨て
- sorted今回いらない。思考途中で入れたままだった。
- 2乗(x-P)**2でいい

## C - String Too Long 【2026/8/3 ADT EASY2】
🔴 **Status: 解説AC**(60分)

```python
N = int(input())
cl = []
C = []
L = []
S = []
ans = []
count = 0

for i in range(N):
  cl = list(map(str, input().split()))
  C.append(cl[0])
  L.append(cl[1])

while sum(map(int, L)) <= 100 and count < N-1:
  for c in C:
    for i in range(int(L[count])):
      ans.append(c)
    count += 1

if sum(map(int, L)) <= 100:
  result = ''.join(ans)
  print(result)
else:
  print("Too Long")
```

- ループの回数が多すぎたことが時間をオーバーする理由だった。  
- result = ''.join(ans)の書き方は調べた。リストから文字列にしている。','とかにすると,で区切れる。
- 解説の解答

```python
N = int(input())
P = []
for i in range(N):
    c, l = input().split()
    P.append([c, int(l)])
if sum(map(lambda x: x[1], P)) > 100:
    print("Too Long")
else:
    S = ""
    for c, l in P:
        S += c * l
    print(S)
```

P.append([c, int(l)])で方が違くても同時に入れている。  
lambda x(引数)['a', 3]のようなペアを取りだし: x[1](処理内容)そのペアの1番目の要素。これにより数字だけを取り出して計算している。  
S = ""で空っぽの文字列を作れる。  
for c, l in P:でぺアを自動で2つの変数にばらして受け取れる。['a', 3]がそのままではなく。  
文字列に数字を掛けえるとその回数分リピートされる。  S +=で新しい文字を後ろにくっつける。

## C - Traveling Salesman around Lake 【Easy 100 #012】
🟢 **Status: 自力AC**(40分)

```python
K, N = map(int, input().split())
befor_a = 0
count = 0
max_a = 0
total = 0

for a in map(int, input().split()):
  if count != 0:
    if max_a < (a - before_a):
      max_a = a - before_a
    total += a - before_a
  before_a = a
  count += 1

if max_a < K - total:
  print(K - K + total)
else:
  print(K- max_a)
```

- ロジックを思いつくのに時間がかかってしまったが詰まったタイミングはなかった

## C - Replacing Integer 【Easy 100 #014】
🟢 **Status: 自力AC**(30分)

```python
N, K = map(int, input().split())
min = 100000000000000000000

while N < min:
  min = N
  N = abs(N - K)

print(min)
```

```python
N, K = map(int, input().split())

if N % K > (K - (N % K)):
  print(K - (N % K))
else:
  print(N % K)
```

- 最初のやつだと1000000000000000000 1の時などに時間がかかりすぎてしまう  
- 絶対値abs
- 紙で計算していく中で基本答えがあまりになっていて、例外としてKからあまりを引いたものがあまりより小さい時答えがKからあまりを引いたものになることが分かったのでそれを利用した

## ABC470C - Inc, Dec, Xor 【2026/8/8 ABC470】
🔴 **Status: 解説AC**(51分)

```python
N, Q = map(int, input().split())
A = [0] * N
Q_list = []

for q in range(Q):
  ans = 0
  Q_list = list(map(int, input().split()))
  if Q_list[0] == 1:
    A[Q_list[1] - 1] += 1
  else:
    for n in range(N):
      if A[n] >= 1:
        A[n] += -1
  for a in A:
    ans ^= a 
  print(ans)
```
