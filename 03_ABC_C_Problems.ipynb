## 🟢C - String Too Long 【2026/8/3 ADT EASY2】
🔴 **Status: 解説AC**(60分)  
🔴 **Status: 解説AC**(40分)  
🟢 **Status: 自力AC**(8分)

```python
N = int(input())
cl = []
C = []
L = []
S = []
ans = []
count = 0

for i in range(N):
  cl = list(map(str, input().split()))
  C.append(cl[0])
  L.append(cl[1])

while sum(map(int, L)) <= 100 and count < N-1:
  for c in C:
    for i in range(int(L[count])):
      ans.append(c)
    count += 1

if sum(map(int, L)) <= 100:
  result = ''.join(ans)
  print(result)
else:
  print("Too Long")
```

- ループの回数が多すぎたことが時間をオーバーする理由だった。  
- <span style="color: red;">result = ''.join(ans)</span>の書き方は調べた。リストから文字列にしている。','とかにすると,で区切れる。
- 解説の解答

```python
N = int(input())
P = []
for i in range(N):
    c, l = input().split()
    P.append([c, int(l)])
if sum(map(lambda x: x[1], P)) > 100:
    print("Too Long")
else:
    S = ""
    for c, l in P:
        S += c * l
    print(S)
```

<span style="color: red;">P.append([c, int(l)])</span>で方が違くても同時に入れている。  
<span style="color: red;">lambda x(引数)</span>['a', 3]のようなペアを取りだし<span style="color: red;">: x[1](処理内容)</span>そのペアの1番目の要素。これにより数字だけを取り出して計算している。  
<span style="color: red;">S = ""</span>で空っぽの文字列を作れる。  
<span style="color: red;">for c, l in P:</span>でぺアを自動で2つの変数にばらして受け取れる。['a', 3]がそのままではなく。  
文字列に数字を掛けえるとその回数分リピートされる。  <span style="color: red;">S +=</span>で新しい文字を後ろにくっつける。  
この解法から新しいことをたくさん学ぶことができた。

```python
N = int(input())
ans = ""
le = 0

for _ in range(N):
    c,l = input().split()
    l = int(l)
    le += l
    if le > 100:
        ans = "Too Long"
        break
    ans += c *l

print(ans)
```

- リストにする必要なかった

## 🟢C - Go to School 【Easy 100 #016】
🔴 **Status: 解説AC**(26分)  
🟢 **Status: 自力AC**(16分)

```python
N = int(input())
A = list(map(int, input().split()))
ans = ""
count = 0

for n in range(1, N+1):
  a = str(A.index(n) + 1)
  ans += a
  count += 1
  if count != N:
    ans += " "

print(ans)
```

- 時間がオーバーする理由が全く分からなかった
- A.index(n)これがN回ループしているのと一緒だった
- <span style="color: red;">list.index(x)</span>で配列の中のxの位置がわかる。ただxがその配列にない場合エラーになる  
- なのでそれを使わない別解

```python
N = int(input())
A = list(map(int, input().split()))

# サイズ N のリストをあらかじめ作っておく
ans = [0] * N

for i in range(N):
  ans[A[i] - 1] = i + 1

# リストの要素をスペース区切りで一括出力
print(*ans)
```

A[i]の要素から1引けば順番になる  
要素は2なら1,3なら2,1なら0  
入る数字はi+1  
<span style="color: red;">*ans</span>でスペース区切りにできる

```python
N = int(input())
count = 1
ans = [0] * N

for A in map(int, input().split()):
    ans[A - 1] = count
    count += 1

print(*ans)
```

```python
import itertools

N = int(input())
V = list(map(int, input().split()))

while len(V) != 1:
  min = 10000000000000
  p =[]
  q = []
  for pair in itertools.combinations(V, 2):
    if sum(pair)/2 < min:
      p, q = map(float, pair)
      min = sum(pair)/2
  V.remove(p)
  V.remove(q)
  V.append(min)

for v in V:
  print(v)
```

## 🟢C - Next Prime 【Easy 100 #021】
🔴 **Status: 解説AC**(30分)  
🟢 **Status: 自力AC**(12分)

```python
def is_prime(n):
  i = 2
  while i**2 <= n:
    if n % i == 0:
      return False
    i += 1
  return True

X = int(input())

while True:
  if is_prime(X):
    print(X)
    break#見つかったら答えを出力して抜ける
  X += 1
```

- TLEになりそうな回答しか思いつかなかった。素数の考え方を知っておくべき問題だと思った。

## 🟢ABC471C - Cookies and Greedy Takahashi 【2026/8/15 ABC471】
🔴 **Status: 解説AC**(84分)  
🔴 **Status: 解説AC**(30分)  
🟢 **Status: 自力AC**(40分)

```python
N = int(input())
A = list(map(int, input().split()))
b = 0
ans = 0

for _ in range(N):
  A=sorted(A, key=lambda x : (abs(x - b), x-b)) 
  ans += abs(A[0] - b)
  b = A[0]
  A.remove(A[0])

print(ans)
```

- TLEになってしまうものしか浮かばなかった
- ❌ループ(N回)の中で毎回sorted()(O(NlogN))とremove()(O(N))を行っているため全体でO(N^2logN)かかる。
- 🔄O(NlogN)+O(N)は近似してO(NlogN)とする。　　  
⏱️計算量は **O(NlogN)** とする
- 別解

```python
N = int(input())
A = list(map(int, input().split()))
A.append(0)
A.sort()

idx = A.index(0)
left = idx -1
right = idx +1

b = 0
ans = 0

for _ in range(N):
  #右側に進めない(右端)
  if right >= len(A):
    target = A[left]#左にしか進めない
    left -= 1
  #左側に進めない(左端)
  elif left < 0:
    target = A[right]
    right += 1
  #左右どちらも選べる
  else:
    dist_left = abs(A[left] - b)
    dist_right = abs(A[right] - b)

    if dist_left <= dist_right:
      target = A[left]
      left -= 1
    else:
      target = A[right]
      right += 1

  ans += abs(target - b)
  b = target

print(ans)
```

N=4の-1 -4 2 -11の場合  
0を入れて-11 -4 -1 0 2  
①right = 2(idx4),left = -1(idx2)    
target = -1,left = -4(idx1)に変わる  
②right = 2(idx4),left = -4(idx1)  
target = -4,left = -11(idx0)に変わる  
③right = 2(idx4),left = -11(idx0)  
target = 2,right = (idx5)に変わる  
④right = (idx5),left = -11(idx0)  
target = -11,left = (idx-1)に変わる  
の流れリストから消す必要もない。

##### ライブラリbisectについて

## 🟢ABC472C - On a Diet 【2026/8/22 ABC472】  
🟢 **Status: 自力AC**(55分)

```python
N, M, K = map(int, input().split())
A = list(map(int, input().split()))

for i in range(N):
    if i < M:
        if sum(A[0:i+1]) > K:
            A[i] = 0
            print("No")
        else:
            print("Yes")
    else:
        if sum(A[(i - M + 1):i+1]) > K:
            A[i] = 0
            print("No")
        else:
            print("Yes")
```

- ❌ループ内O(N)のA[(i - M + 1):i+1]が:(スライス)を使っていて、これは毎回その範囲の要素をコピーして新しいリストを作成これで計算量がO(M)さらに、そのリストの先頭から順に足しているのでsum()も計算量O(M)かかっている。ここで2Mつまり計算量O(M)となる。Mに比例しているので。  
⏱️つまり全体の**計算量はO(N×M)**

```python
N, M, K = map(int, input().split())
total = 0
count = 0
a_mem = []

for a in map(int, input().split()):
    if count >= M:
        total -= a_mem[count-M]
    if (total + a) > K:
        print("No")
        a_mem.append(0)
    else:
        total += a
        print("Yes")
        a_mem.append(a)
    count += 1
```

- 🔄TLEになる原因であったsumとスライスをなくし、ループの中の計算量をO(1)に。直前の和からの差分(追加と削除)だけを管理するようにしている。尺取り法
