## 🔴ABC479D - Back and Forth Distance【2026/10/11 ABC479】
🔴 **Status: 解説AC**(65分)

```python
import sys

def solve():
    input = sys.stdin.read#一括で読み込む関数をinputに置きかえた
    data = input().split()
    if not data:
        return

    N = int(data[0])
    Q = int(data[1])
    
    L = [int(x) for x in data[2:2+N-1]]
    R = [int(x) for x in data[2+N-1:2+2*(N-1)]]
    
    # 1-indexedで扱いやすくするための配列調整
    cost_L = [0]*2+L
    cost_R = [0] + R + [0]
    
    # プレフィックス最小値: 1 から i-1 までの cost_R の最小値
    pref_min_R = [float("inf")]*(N+1)#マス1からNまでに対応するリスト(1-indexedで扱いやすく)
    curr = float("inf")
    for i in range(1,N+1):
        pref_min_R[i] = curr
        if i < N:
            curr = min(curr,cost_R[i])
    
    # サフィックス最小値: i+1 から N までの cost_L の最小値
    suff_min_L = [float("inf")]*(N+1)
    curr = float("inf")
    for i in range(N,0,-1):
        suff_min_L[i] = curr
        if i > 1:
            curr = min(curr, cost_L[i])
    
    #クエリを1つずつ順番に効率よく取り出す
    results = []#クエリごとの計算結果をリストに入れ後でまとめて出力
    q_idx = 2 + 2*(N-1)
    for _ in range(Q):
        s = int(data[q_idx])
        t = int(data[q_idx+1])
        q_idx += 2
        
        if s < t:
            ans = cost_R[s]
            ans = min(ans, cost_L[s]+pref_min_R[s])
            #右へ行きすぎてから戻るこれが抜けていた
            ans = min(ans, cost_R[s]+suff_min_L[t])
        else:
            ans = cost_L[s]
            ans = min(ans, cost_R[s]+suff_min_L[s])
            #左へ行きすぎてから戻るこれも抜けていた
            ans = min(ans, cost_L[s]+pref_min_R[t])
        
        results.append(str(ans))
    
    print("\n".join(results))#join()はリストの中身が文字列である必要がある
    

if __name__ == "__main__":
  solve()
```

- 🔄⏱️あらかじめ累積最小値（プレフィックス min とサフィックス min）を使って計算しておくことで、各クエリをO(1)で処理できる

##### 📚ライブラリsysについて  
D問題以上、または制約が10^5以上の問題で使った方がいいテクニック  
入力を一括でできる

```python
import sys

def solve():
    input = sys.stdin.read#一括で読み込む関数をinputに置きかえた
    data = input().split()
    if not data:#入力データが空っぽの時プログラム終了
        return

if __name__ == "__main__":
  solve()
```
