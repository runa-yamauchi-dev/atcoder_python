## 🟢B03 - Supermarket 1 
🟢 **Status: 自力AC**(11分)

```python
N = int(input())
A = list(map(int, input().split()))
ans = "No"

for a1 in A:
    for a2 in A:
        if a1 == a2:
            break
        else:
            for a3 in A:
                if a1 == a3 or a2 == a3:
                    break
                else:
                    if 1000-(a1+a2) == a3:
                        ans = "Yes"
                        break

print(ans)
```

- 解答例(こっちの方がきれい)

```python
N = int(input())
A = list(map(int, input().split()))
ans = "No"

for i in range(N):
	for j in range(i+1, N):
		for k in range(j+1, N):
			if A[i] + A[j] + A[k] == 1000:
				ans = "Yes"

print(ans)
```

これだと重複が起こらず、すべてを確かめることができる

## 🟠A05 - Three Cards 
🟠 **Status: 実装もたつきAC**(50分)

```python
N, K = map(int, input().split()) 
NUM = list(range(1,N+1))
count = 0

for i in range(N-1,-1,-1):
    if K - NUM[i] <= K:
        NUM_1 = K - NUM[i]
        for j in range(N-1,-1,-1):
            if 0 < NUM_1 - NUM[j] <= N:
                count += 1

print(count)
```

- ロジックを思いつくのにかなり時間がかかったがなんとか行けた　　
- 解答のロジックと比較しても考え方が近くてよかった

```python
N, K = map(int, input().split())
count = 0

for x in range(1,N+1):
    for y in range(1,N+1):
        z = K - x -y
        if 1 <= z <= N:
            count += 1

print(count)
```

- これが解答例だが計算量は自分の奴のほうが短かった
- しかしこっちのほうが可読性が高くバグを生みにくいので良い
- 🔄たとえば2つの値が分かっていれば自動的にもう1つの値が決まる場合ループを削減することができる  
⏱️計算量を **O(N^3→N^2)** にできる

## 🟢B06 - Lottery 
🟢 **Status: 自力AC**(10分)

```python
N = int(input())
A = [0]
A_total = 0

for a in map (int, input().split()):
    A_total += a
    A.append(A_total)

for _ in range(int(input())):
    L, R = map(int, input().split())
    if 2*(A[R]-A[L-1]) > (R-L+1):
        print("win")
    elif 2*(A[R]-A[L-1]) == (R-L+1):
        print("draw")
    else:
        print("lose")
```

## 🔴A07 - Event Attendance 
🔴 **Status: 解説AC**(28分)

```python
D = int(input())
N = int(input())
A = [0]*D
ans = 0

for _ in range(N):
    L, R = map(int, input().split())
    A[L-1] += 1
    if R < D:
        A[R] -= 1

for a in A:
    ans += a
    print(ans)
```

- そもそも自分で考えたロジック計算合わないものだった4秒までの繰り返しかと思っていたが6秒の時など例外が存在した

```python
def check(A, M, K):
    sum_M = 0
    for a in A:
        sum_M += (M // a)
    if sum_M >= K:
        return True#ここで関数抜ける(関数はreturnで抜ける)
    return False

N, K = map(int, input().split())
A = list(map(int, input().split()))

L = 1
R = 10**9

while L != R:
    M = (L+R)//2
    if check(A, M, K):
        R = M
    else:
        L = M+1

print(L)
```

- ⏱️関数の中の計算量がO(N)、**二分探索はO(logN)** めっちゃ小さい

## 第三章まとめ
#### 二分探索　　
中央で区切ることで探索範囲を半分ずつにしていくアルゴリズム  
例えば1~64で数字をYes/Noのみで当てるとき64は2の6乗なので6回で答えを求められる  
答えが〇以下ですかとかと判定問題がこたえられるとき使える。単調増加・減少のものなど

- 二分探索は半分に割って
- 尺取りは1つずつ隣

##### 📚**二分探索はこのルールを覚える**  

```python
N = int(input())
A = sorted(list(map(int, input().split())))
Q = int(input())

for _ in range(Q):
    X = int(input())
    L = 0
    R = N
    while L < R: 
        M = (L+R)//2
        if X <= A[M]:
            R = M
        else:
            L = M+1
    print(L)
```

- sortしてあるものを使う
- 初期値L = 0, `R = N`で成り立つとき`R = M`,成り立たないとき`L = M+1`を使う
- M = (L+R)//2
- `while L < R:`のほうが安全
- `以上(idx = bisect.bisect_left(A, X))`と`それより大きい(idx = bisect.bisect_right(A, X))`A[M] >= **X**とA[M] > **X**
