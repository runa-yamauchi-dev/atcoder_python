## 🟢A02 - Linear Search 
🟢 **Status: 自力AC**(3分) 

```python
N, X = map(int, input().split())
ans = "No"

for a in map(int, input().split()):
    if a == X:
        ans = "Yes"
        break

print(ans)
```

## 🟢B02 - Divisor Check 
🟢 **Status: 自力AC**(5分) 

```python
A, B = map(int, input().split())
ans = "No"

for i in range(A,B+1):
    if 100 % i == 0:
        ans = "Yes"
        break

print(ans)
```

## 🟢A03 - Two Cards 
🟢 **Status: 自力AC**(7分) 

```python
N, K = map(int, input().split())
P = list(map(int, input().split()))
Q = list(map(int, input().split()))
ans = "No"

for p in P:
    for q in Q:
        if p+q == K:
            ans = "Yes"
            break

print(ans)
```

- 全探索で計算量はO(N^2)今回N=100のため10000となるので間に合う

## 🟢B03 - Supermarket 1 
🟢 **Status: 自力AC**(11分)

```python
N = int(input())
A = list(map(int, input().split()))
ans = "No"

for a1 in A:
    for a2 in A:
        if a1 == a2:
            break
        else:
            for a3 in A:
                if a1 == a3 or a2 == a3:
                    break
                else:
                    if 1000-(a1+a2) == a3:
                        ans = "Yes"
                        break

print(ans)
```

- 解答例(こっちの方がきれい)

```python
N = int(input())
A = list(map(int, input().split()))
ans = "No"

for i in range(N):
	for j in range(i+1, N):
		for k in range(j+1, N):
			if A[i] + A[j] + A[k] == 1000:
				ans = "Yes"

print(ans)
```

これだと重複が起こらず、すべてを確かめることができる

## 第一章まとめ
#### 全探索　　
あり得るすべてのパターンを調べあげるという方針で問題を解く手法  
  
**線形探索法**...データを先頭から順に目的のデータと一致するかを確認していく。計算量O(N)
