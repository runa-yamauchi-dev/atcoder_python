## 🟢A03 - Two Cards 
🟢 **Status: 自力AC**(7分) 

```python
N, K = map(int, input().split())
P = list(map(int, input().split()))
Q = list(map(int, input().split()))
ans = "No"

for p in P:
    for q in Q:
        if p+q == K:
            ans = "Yes"
            break

print(ans)
```

## 🟢B03 - Supermarket 1 
🟢 **Status: 自力AC**(11分)

```python
N = int(input())
A = list(map(int, input().split()))
ans = "No"

for a1 in A:
    for a2 in A:
        if a1 == a2:
            break
        else:
            for a3 in A:
                if a1 == a3 or a2 == a3:
                    break
                else:
                    if 1000-(a1+a2) == a3:
                        ans = "Yes"
                        break

print(ans)
```

- 解答例(こっちの方がきれい)

```python
N = int(input())
A = list(map(int, input().split()))
ans = "No"

for i in range(N):
	for j in range(i+1, N):
		for k in range(j+1, N):
			if A[i] + A[j] + A[k] == 1000:
				ans = "Yes"

print(ans)
```

これだと重複が起こらず、すべてを確かめることができる

## 🟢A04 - Binary Representation 1 
🟢 **Status: 自力AC**(22分)

```python
N = int(input())
ans = ""

for _ in range(10):
    if N != 0:
        ans += str(N % 2)
        N //= 2
    else:
        ans += "0"

print(ans[::-1])
```

## 🟢B04 - Binary Representation 2

```python
N = input()
l = len(N) - 1 
ans = 0

for n in N:
    ans += int(n)*(2**l)
    l -= 1

print(ans)  
```

## 🟠A05 - Three Cards 
🟠 **Status: 実装もたつきAC**(50分)

```python
N, K = map(int, input().split()) 
NUM = list(range(1,N+1))
count = 0

for i in range(N-1,-1,-1):
    if K - NUM[i] <= K:
        NUM_1 = K - NUM[i]
        for j in range(N-1,-1,-1):
            if 0 < NUM_1 - NUM[j] <= N:
                count += 1

print(count)
```

- ロジックを思いつくのにかなり時間がかかったがなんとか行けた　　
- 解答のロジックと比較しても考え方が近くてよかった

```python
N, K = map(int, input().split())
count = 0

for x in range(1,N+1):
    for y in range(1,N+1):
        z = K - x -y
        if 1 <= z <= N:
            count += 1

print(count)
```

- これが解答例だが計算量は自分の奴のほうが短かった
- しかしこっちのほうが可読性が高くバグを生みにくいので良い
- 🔄たとえば2つの値が分かっていれば自動的にもう1つの値が決まる場合ループを削減することができる  
⏱️計算量を **O(N^3→N^2)** にできる

## 🟢A06 - How Many Guests? 
🟢 **Status: 自力AC**(11分)

```python
N, Q = map(int, input().split())
A = [0]
total = 0

for a in map(int, input().split()):
    total += a
    A.append(total)

for _ in range(Q):
    L, R = map(int, input().split())
    print(A[R] - A[L-1])
```

## 🟢B06 - Lottery 
🟢 **Status: 自力AC**(10分)

```python
N = int(input())
A = [0]
A_total = 0

for a in map (int, input().split()):
    A_total += a
    A.append(A_total)

for _ in range(int(input())):
    L, R = map(int, input().split())
    if 2*(A[R]-A[L-1]) > (R-L+1):
        print("win")
    elif 2*(A[R]-A[L-1]) == (R-L+1):
        print("draw")
    else:
        print("lose")
```

## 🔴A11 - Binary Search 1 
🔴 **Status: 解説AC**(32分)

```python
N, X = map(int, input().split())
A = list(map(int, input().split()))

L = 1
R = N

while True:
    M = (L+R)//2
    
    if X < A[M-1]:
        R = M-1
    elif X > A[M-1]:
        L = M+1
    else:
        print(M)
        break
```

- bisectライブラリを使った例も   

```python
import bisect

# 入力の受け取り
N, X = map(int, input().split())
A = list(map(int, input().split()))

# A はすでに昇順にソートされているので、そのまま bisect_left を使用
# X 以上の要素が最初に現れるインデックス（0始まり）を取得する
idx = bisect.bisect_left(A, X)

# 問題の出力指定は「1番目始まり（1-indexed）」なので +1 する
print(idx+1)
```

```python
import bisect

N, X = map(int, input().split())
A = list(map(int, input().split()))

# X より大きいの要素が最初に現れるインデックス（0始まり）を取得する
idx = bisect.bisect_right(A, X)

print(idx)
```
