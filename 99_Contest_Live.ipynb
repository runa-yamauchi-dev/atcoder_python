### B問題:

```python
N = int(input())
S = input()
position = []
newposition = []
count = 0
num = 0

for s in S:
  if s == "x":
    position.append(count)
  count += 1

if len(S) > 1:
  for p in position:
    if p == 0:
      if S[p+1] == "o":
        newposition.append(p)
    elif p == N-1:
      if S[p-1] == "o":
        newposition.append(p)
    else:
      if S[p-1] == "o" or S[p+1] == "o":
        newposition.append(p)
    
print(len(position)-len(newposition))

```

### 📝振り返り  

- 最初position.remove(p)でやっていてうまくいかなかったのはリストをループしている途中で列の中から人を抜き取ってしまったことで抜き取った次の人を飛ばしてしまっていた。position = [0, 2, 4]のあとインデックス0のp=0を消した場合次インデックス1を見ようとするがposition = [2, 4]となっているのでp=4を見ていることになる。P=2を飛ばしている。  
- forループでremove()は使わない  
- if len(S) > 1:はエラーから自分で思いついてかいた
- もう一個リストを作るというのは調べていく中で発想を得た。消すのが無理だったので逆に増やすという発想。  
- そこからもう一個のリストを消すべき要素として長さを引くというロジックを思いつけた。  
- 別解  


```python
N = int(input())
S = input()

ans = 0
for i in range(N):#N回ループ
    if S[i] == "x":
        # 左に o があるかチェック（自分が左端でない場合）
        has_left_o = (i > 0 and S[i - 1] == "o")
        
        # 右に o があるかチェック（自分が右端でない場合）
        has_right_o = (i < N - 1 and S[i + 1] == "o")
        
        # どちらにも o がない場合（両方oでない）をカウント
        if not (has_left_o or has_right_o):
            ans += 1

print(ans)
```

has_left_oなどにはTrueかFalseが入っている

```python
S, A, B, X = map(int, input().split())
ans = 0
n = 0
count_A = 0

while X != 0:
  if X >= A:
    X -= A
    count_A += 1
  else:
    n = X
    break
  if X >= B:
    X -= B
  else:
    break

ans = (S * count_A * A) + (S * n) 

print(ans) 
```

```python
N = int(input())
cl = []
C = []
L = []
S = []
ans = []
count = 0

for i in range(N):
  cl = list(map(str, input().split()))
  C.append(cl[0])
  L.append(cl[1])

while sum(map(int, L)) <= 100 and count < N-1:
  for c in C:
    for i in range(int(L[count])):
      ans.append(c)
    count += 1

if sum(map(int, L)) <= 100:
  result = ''.join(ans)
  print(result)
else:
  print("Too Long")
```

### 📝振り返り  

- ループの回数が多すぎたことが時間をオーバーする理由だった。  
- result = ''.join(ans)の書き方は調べた。リストから文字列にしている。','とかにすると,で区切れる。
- 解説の解答

```python
N = int(input())
P = []
for i in range(N):
    c, l = input().split()
    P.append([c, int(l)])
if sum(map(lambda x: x[1], P)) > 100:
    print("Too Long")
else:
    S = ""
    for c, l in P:
        S += c * l
    print(S)
```

P.append([c, int(l)])で方が違くても同時に入れている。  
lambda x(引数)['a', 3]のようなペアを取りだし: x[1](処理内容)そのペアの1番目の要素。これにより数字だけを取り出して計算している。  
S = ""で空っぽの文字列を作れる。  
for c, l in P:でぺアを自動で2つの変数にばらして受け取れる。['a', 3]がそのままではなく。  
文字列に数字を掛けえるとその回数分リピートされる。  S +=で新しい文字を後ろにくっつける。
