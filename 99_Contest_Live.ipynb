### A問題:

```python
S = input()
ans = ""

for s in S:
    if s == "A":
        ans += s
    else:
        ans += "."

print(ans)
```

### B問題:

```python
N = int(input())
L = list(map(int, input().split()))
le = sum(L)
ls = 0
l_min = 1000000
count = 0

for l in L:
    if count == (N - 1):
        break
    le -= l
    ls += l 
    l_min = min(l_min, abs(le-ls))
    count += 1

print(l_min)
```

### C問題:

```python
N, M, K = map(int, input().split())
A = list(map(int, input().split()))

for i in range(N):
    if i < M:
        if sum(A[0:i+1]) > K:
            A[i] = 0
            print("No")
        else:
            print("Yes")
    else:
        if sum(A[(i - M + 1):i+1]) > K:
            A[i] = 0
            print("No")
        else:
            print("Yes")
```

```python
N, M, K = map(int, input().split())
total = 0
count = 0
a_mem = []

for a in map(int, input().split()):
    if count >= M:
        total -= a_mem[count-M]
    if (total + a) > K:
        print("No")
        a_mem.append(0)
    else:
        total += a
        print("Yes")
        a_mem.append(a)
    count += 1
```
