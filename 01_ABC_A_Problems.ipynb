## 🟢A - Cookie Exchanges 【Easy 100 #013】 
🟢 **Status: 自力AC**(13分)

```python
A, B, C = map(int, input().split())
count = 0

while (A % 2 == 0) and (B % 2 == 0) and (C % 2 == 0):
  if A == B == C:
    count = -1
    break
  a = A
  b = B
  c = C
  A = (b // 2) + (c // 2)
  B = (a // 2) + (c // 2)
  C = (a // 2) + (b // 2)
  count += 1

print(count)
```

## 🟢A - Candy Distribution Again 【Easy 100 #022】
🟢 **Status: 自力AC**(15分)

```python
N, x = map(int, input().split())
ans = 0


for a in sorted(map(int, input().split())):
  if x >= a:
    if ans == N - 1 and x != a:
      break
    ans += 1
    x -= a
  
print(ans)
```

## 🟢ABC471A - Nine or Nein 【2026/8/15 ABC471】
🟢 **Status: 自力AC**(3分)

```python
A, B = map(int, input().split())
if (A + B == 9) or (A - B == 9) or (A * B == 9) or (A / B == 9):
  print("Nine")
else:
  print("Nein")
```

## 🟢A - Range Product 【Easy 100 #040】 
🟢 **Status: 自力AC**(15分)

```python
a, b = map(int, input().split())

if a <= 0 and b >= 0:
    print("Zero")
elif a > 0 and b > 0:
    print("Positive")
else:
    if (abs(a)-abs(b)+1) % 2 == 0:
        print("Positive")
    else:
        print("Negative")
```

- abs(a)-abs(b)+1はb-a+1でいい

## 🟢ABC473A - Second Half Sum【2026/8/29 ABC473】
🟢 **Status: 自力AC**(4分)

```python
N = int(input())
count = 1
ans = 0

for a in map(int, input().split()):
    if count >= N/2+1:
        ans += a
    count +=1
    
print(ans)
```
