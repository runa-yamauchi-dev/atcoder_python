```python
N = int(input())
count_l = 0
ans = 1

for n in range(1,N+1,1):
  count = 0
  num = n
  if n % 2 == 0:
    while n % 2 == 0:
      n /= 2
      count += 1
  if count_l < count:
    count_l = count
    ans = num

print(ans)
```

## 🟢B - Collatz Problem 【Easy 100 #020】
🟢 **Status: 自力AC**(20分)

```python
s = int(input())
a = s
a_lis = []
a_lis.append(a)
ans = 0
count = 1

while True:
  if a % 2 == 0:
    a /= 2
  else:
    a = (3*a) + 1 
  count += 1
  if a in a_lis:
    ans = count
    break
  a_lis.append(a)

print(ans)
```

## 🟢B - Count Balls 【Easy 100 #027】
🟠 **Status: 実装もたつきAC**(35分)  
🟢 **Status: 自力AC**(16分)

```python
N, A, B = map(int, input().split())
ans = 0

if N > A and N < (A + B):
  ans = A
else:
  if N % (A + B) > A:
    ans += A * (N // (A + B)) + A
  else:
    ans += A * (N // (A + B)) + (N % (A + B))

print(ans)
```

- 🤔一応ACにはできたがぬけが多くて修正するのに時間がかかってしまった  
- 🔄なので以下のようにN個の中に「青A個＋赤B個」のセットが何回入るかを 商 // と 余り % で分解し、余りとAをminで比較するとよい

```python
N, A, B = map(int, input().split())

print(A * (N // (A+B)) + min((N % (A+B)), A))
```

```python
S = input()
Alphabet = ["a", "b", "c", "d", "e", "f", "g", "h", "i", "j", "k", "l", "m", "n", "o", "p", "q", "r", "s", "t", "u", "v", "w", "x", "y", "z"]
ans = "None"
le = len(S)

for a in Alphabet:
    count = 0
    for s in S:
        if s == a:
            break
        count += 1
    if count == le:
        ans = a
        break

print(ans)
```

- 計算量的に問題なさそうだったので全探索でやった

```python
import string

S_set = set(input())  # 検索を O(1) にするために set に変換

for a in string.ascii_lowercase:
    if a not in S_set:
        print(a)
        break
else:
    print("None")
```

- 📚<span style="color: red;">a not in S_set(文字列でもsetでもok)</span>の部分S_setにaが含まれていないかを判定inの前にnot  
- 📚<span style="color: red;">import string</span>によって<span style="color: red;">string.ascii_lowercase</span>というアルファベットの配列が使える
- 📚<span style="color: red;">for-else構文</span>...forと同階層にelseを書ける

##### 📚ライブラリstringについて  
- string.ascii_lowercase...英小文字  
- string.ascii_uppercase...英大文字  
- string.ascii_letters...大文字小文字どっちも  
- string.digits...0~9の数字  
- string.hexdigits...16進数0~9,a~f,A~F  
- string.punctuation...記号
