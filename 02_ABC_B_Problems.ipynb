```python
A, B = map(int, input().split())
a = 1
b = 0
ans = 0

if B == 1:
  print(0)
else:
  if A >= B:
    print(1) 
  else:
    ans = ((B-1)/(A-1))
    if (B-1)%(A-1) == 0:
      print(int(ans))
    else:
      print(1+int(ans))
```
