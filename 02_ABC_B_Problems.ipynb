## B - Qualification simulator 【Easy 100 #003】
🟢 **Status: 自力AC**(20分)

```python
N, A, B = map(int, input().split())
S = input()
count = 0
count_b = 0

for s in S:
  if s == "a":
    if (A + B) <= count:
      print("No")
    else:
      print("Yes")
      count += 1
  elif s == "b":
    if (A + B) <= count or B <= count_b:
      print("No")
    else:
      print("Yes")
      count_b += 1
      count += 1
  else:
    print("No")
```

- もたつきはなかったが、(A + B)の()とかつけるの忘れてて時間がBにしてはかかったかも。詰まったタイミングはなかった。
