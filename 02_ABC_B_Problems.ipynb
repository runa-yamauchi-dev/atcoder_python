## B - Power Socket 【Easy 100 #001】
🟠 **Status: 実装もたつきAC**(時間)  
🟢 **Status: 自力AC**(12分)

```python
A, B = map(int, input().split())
ans = 0

if B == 1:
  print(0)
else:
  if A >= B:
    print(1) 
  else:
    ans = ((B-1)/(A-1))
    if (B-1)%(A-1) == 0:
      print(int(ans))
    else:
      print(1+int(ans))
```

- 余りを使って小数か整数かを判断
- ロジックを考えるのに少し時間がかかった。1回全部消して考え直した。

```python
A, B = map(int, input().split())
n = 0
ans = 1

if A < B:
  n = B - A

if n % (A - 1) == 0:
  ans += (n // (A - 1))
else:
  ans += ((n // (A - 1)) + 1)

if B == 1:
  ans = 0

print(ans)
```

## B - Qualification simulator 【Easy 100 #003】
🟢 **Status: 自力AC**(20分)

```python
N, A, B = map(int, input().split())
S = input()
count = 0
count_b = 0

for s in S:
  if s == "a":
    if (A + B) <= count:
      print("No")
    else:
      print("Yes")
      count += 1
  elif s == "b":
    if (A + B) <= count or B <= count_b:
      print("No")
    else:
      print("Yes")
      count_b += 1
      count += 1
  else:
    print("No")
```

## B - Tax Rate 【Easy 100 #004】
🔴 **Status: 解説AC**  
🟢 **Status: 自力AC**(20分)


```python
N = int(input())

if int(int(((N * 100) / 108) + 0.5) * 1.08) == N:
  print(int(((N * 100) / 108) + 0.5))
elif int((int(((N * 100) / 108) + 0.5) + 1) * 1.08) == N:
  print(int(((N * 100) / 108) + 0.5) + 1)
else:
  print(":(")
```

```python
N = int(input())

if int((N // 1.08) * 1.08) == N:
  print(int(N // 1.08)) 
elif int(((N // 1.08) + 1) * 1.08) == N:
  print(int(N // 1.08) + 1)
else:
  print(":(")
```

- 今回は方針がよくなかったのと。WAの理由を見つけるのが不可能に近かった。普通に例から考えた場合、Nを1.08で割って四捨五入したものがXだと思ってしまった。だが反例があった。たとえばN＝25の場合25÷1.08=23.1...で23×1.08=24.84。しかしX=24が24×1.08=25.92より24ならいける。つまり25.92のように切り捨てる前の金額がギリギリの場合四捨五入した値+1で成り立ってしまう。  
- ただこれを自力で見つけるのはかなり難しい。今回のようにループが5万回程度で全部で10^8より小さい場合全部試した方が無駄なミスがなくなりいい。

```python
N = int(input())
ans = 0

for X in range(1, 50001):
  if int(X * 1.08) == N:
    print(X)
    ans = 1
    break    

if ans == 0:
  print(":(")
```

## B - Bishop 【Easy 100 #006】

```python
H, W = map(int, input().split())

if (H == 1) or (W == 1):
  print(1)
elif (H * W) % 2 == 0:
  print(int((H * W) / 2))
else:
  print(int((H * W) // 2 + 1))
```

## B - Bingo 【Easy 100 #007】
🔴 **Status: 解説AC**(40分)  
🟢 **Status: 自力AC**(35分)  
もう一度解く

```python
A = [list(map(int, input().split())) for _ in range(3)]
N = int(input())
ans = "No"
for n in range(N):
  b = int(input())
  for i in range(3):
    for j in range(3):
      if A[i][j] == b:
        A[i][j] = ""

for i in range(3):
  if A[i][0] == A[i][1] == A[i][2] == "":
    ans = "Yes"
    break
  elif A[0][i] == A[1][i] == A[2][i] == "":
    ans = "Yes"
    break
  elif (A[0][0] == A[1][1] == A[2][2] == "") or (A[2][0] == A[1][1] == A[0][2] == ""):
    ans = "Yes"
    break

print(ans)
```

- 例の検証に時間がかかってしまって時間が長かったがロジック自体は自力で思いつくことができた。ただ今回は初めてのミスのパターンで出力の文字の大文字小文字を間違えていた。これは本番気づける気がしないので頭に入れておく。

- A = [A1, A2, A3]とまとめてよかった

## B - Grandma's Footsteps 【2026/8/3 ADT EASY2】
🟢 **Status: 自力AC**(20分)

```python
S, A, B, X = map(int, input().split())
ans = 0
n = 0
count_A = 0

while X != 0:
  if X >= A:
    X -= A
    count_A += 1
  else:
    n = X
    break
  if X >= B:
    X -= B
  else:
    break

ans = (S * count_A * A) + (S * n) 

print(ans) 
```

## B - Break Number 【Easy 100 #011】
🟢 **Status: 自力AC**(20分)

```python
N = int(input())
count_l = 0
ans = 1

for n in range(1,N+1,1):
  count = 0
  num = n
  if n % 2 == 0:
    while n % 2 == 0:
      n /= 2
      count += 1
  if count_l < count:
    count_l = count
    ans = num

print(ans)
```

- range(1,N+1,1)は(1, N+1)だけで平気  
- 一番多く割れるN以下の数はN以下の最大の2の累乗(1,2,4,8など)なのでNを超えない間ansにひたすら2を掛けるのもあり

## ABC470B - Monocolor 【2026/8/8 ABC470】
🟢 **Status: 自力AC**(12分)

```python
N = int(input())
C = list(map(int, input().split()))
count = 0
num = 0
ans = 0

for n in range(1, N+1):
  if count < C.count(n):
    count = C.count(n)
    num = n

for c in C:
  if num != c:
    ans += 1

print(ans)
```
