## B - Qualification simulator 【Easy 100 #003】
🟢 **Status: 自力AC**(20分)

````python
N, A, B = map(int, input().split())
S = input()
count = 0
count_b = 0

for s in S:
  if s == "a":
    if (A + B) <= count:
      print("No")
    else:
      print("Yes")
      count += 1
  elif s == "b":
    if (A + B) <= count or B <= count_b:
      print("No")
    else:
      print("Yes")
      count_b += 1
      count += 1
  else:
    print("No")
````

- もたつきはなかったが、(A + B)の()とかつけるの忘れてて時間がBにしてはかかったかも。詰まったタイミングはなかった。

## B - Tax Rate 【Easy 100 #004】
🔴 **Status: 解説AC**

````python
N = int(input())

if int(int(((N * 100) / 108) + 0.5) * 1.08) == N:
  print(int(((N * 100) / 108) + 0.5))
elif int((int(((N * 100) / 108) + 0.5) + 1) * 1.08) == N:
  print(int(((N * 100) / 108) + 0.5) + 1)
else:
  print(":(")
````

- 今回は方針がよくなかったのと。WAの理由を見つけるのが不可能に近かった。普通に例から考えた場合、Nを1.08で割って四捨五入したものがXだと思ってしまった。だが反例があった。たとえばN＝25の場合25÷1.08=23.1...で23×1.08=24.84。しかしX=24が24×1.08=25.92より24ならいける。つまり25.92のように切り捨てる前の金額がギリギリの場合四捨五入した値+1で成り立ってしまう。  
- ただこれを自力で見つけるのはかなり難しい。今回のようにループが5万回程度で全部で10^8より小さい場合全部試した方が無駄なミスがなくなりいい。

## B - Can you solve this? 【Easy 100 #005】
🟢 **Status: 自力AC**(14分)

````python
N, M ,C = map(int, input().split())
B = list(map(int, input().split()))
count = 0

for n in range(N):
  A = list(map(int, input().split()))
  sosu = C
  for m in range(M):
    sosu += A[m] * B[m]
  if sosu > 0:
    count += 1

print(count)
````

## B - Bishop 【Easy 100 #006】
🟢 **Status: 自力AC**(25分)

````python
H, W = map(int, input().split())

if (H == 1) or (W == 1):
  print(1)
elif (H * W) % 2 == 0:
  print(int((H * W) / 2))
else:
  print(int((H * W) // 2 + 1))
````

- 結果を待つ時間入れると30分くらい。ロジックは問題文を理解してからは割とすぐ思い浮かんだ。  
- 紙に書いてたら規則性を見つけることができたのでそれを使った
- HまたはWが1の時だけ変わるというのを思いつくのに少し時間がかかった。
- 10^9と制約があったため全部試すのはやめた

## B - Bingo 【Easy 100 #007】
🔴 **Status: 解説AC**(40分)

````python
A1 = list(map(int, input().split()))
A2 = list(map(int, input().split()))
A3 = list(map(int, input().split()))
N = int(input())
list_a1 = []
list_a2 = []
list_a3 = []
ans = "No"

for n in range(N):
  b = int(input())
  count = 0
  for a1 in A1:
    if a1 == b:
      list_a1.append(count)
    count += 1 
  count = 0
  for a2 in A2:
    if a2 == b:
      list_a2.append(count)
    count += 1
  count = 0
  for a3 in A3:
    if a3 == b:
      list_a3.append(count)
    count += 1

if len(list_a1) == 3 or len(list_a2) == 3  or len(list_a3) == 3:
  ans = "Yes"

for i in range(3):
  if int(i in list_a1) == 1 and int(i in list_a2) == 1 and int(i in list_a3) == 1:
    ans = "Yes"

if (int(0 in list_a1) == 1 and int(1 in list_a2) == 1 and int(2 in list_a3) == 1) or (int(2 in list_a1) == 1 and int(1 in list_a2) == 1 and int(0 in list_a3) == 1):
  ans = "Yes"

print(ans)

````
