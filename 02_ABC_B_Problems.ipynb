## B - Qualification simulator 【Easy 100 #003】
🟢 **Status: 自力AC**(20分)

```python
N, A, B = map(int, input().split())
S = input()
count = 0
count_b = 0

for s in S:
  if s == "a":
    if (A + B) <= count:
      print("No")
    else:
      print("Yes")
      count += 1
  elif s == "b":
    if (A + B) <= count or B <= count_b:
      print("No")
    else:
      print("Yes")
      count_b += 1
      count += 1
  else:
    print("No")
```

- もたつきはなかったが、(A + B)の()とかつけるの忘れてて時間がBにしてはかかったかも。詰まったタイミングはなかった。

## B - Tax Rate 【Easy 100 #004】
🔴 **Status: 解説AC**

```python
N = int(input())

if int(int(((N * 100) / 108) + 0.5) * 1.08) == N:
  print(int(((N * 100) / 108) + 0.5))
elif int((int(((N * 100) / 108) + 0.5) + 1) * 1.08) == N:
  print(int(((N * 100) / 108) + 0.5) + 1)
else:
  print(":(")
```

- 今回は方針がよくなかったのと。WAの理由を見つけるのが不可能に近かった。普通に例から考えた場合、Nを1.08で割って四捨五入したものがXだと思ってしまった。だが反例があった。たとえばN＝25の場合25÷1.08=23.1...で23×1.08=24.84。しかしX=24が24×1.08=25.92より24ならいける。つまり25.92のように切り捨てる前の金額がギリギリの場合四捨五入した値+1で成り立ってしまう。  
- ただこれを自力で見つけるのはかなり難しい。今回のようにループが5万回程度で全部で10^8より小さい場合全部試した方が無駄なミスがなくなりいい。

## B - Bishop 【Easy 100 #006】
🟢 **Status: 自力AC**(25分)

```python
H, W = map(int, input().split())

if (H == 1) or (W == 1):
  print(1)
elif (H * W) % 2 == 0:
  print(int((H * W) / 2))
else:
  print(int((H * W) // 2 + 1))
```

## B - Collecting Balls (Easy Version) 【Easy 100 #009】
🟢 **Status: 自力AC**(15分)

```python
N = int(input())
K = int(input())
ans = 0

for x in map(int, input().split()):
  if x < (K - x):
    ans += x * 2
  else:
    ans += (K - x) * 2

print(ans)
```

## ABC469B - Isolated Seats 【2026/8/1 ABC469】
🟠 **Status: 実装もたつきAC**(90分)

```python
N = int(input())
S = input()
position = []
newposition = []
count = 0
num = 0

for s in S:
  if s == "x":
    position.append(count)
  count += 1

if len(S) > 1:
  for p in position:
    if p == 0:
      if S[p+1] == "o":
        newposition.append(p)
    elif p == N-1:
      if S[p-1] == "o":
        newposition.append(p)
    else:
      if S[p-1] == "o" or S[p+1] == "o":
        newposition.append(p)
    
print(len(position)-len(newposition))
```

- 最初position.remove(p)でやっていてうまくいかなかったのはリストをループしている途中で列の中から人を抜き取ってしまったことで抜き取った次の人を飛ばしてしまっていた。position = [0, 2, 4]のあとインデックス0のp=0を消した場合次インデックス1を見ようとするがposition = [2, 4]となっているのでp=4を見ていることになる。P=2を飛ばしている。  
- forループでremove()は使わない  
- if len(S) > 1:はエラーから自分で思いついてかいた
- もう一個リストを作るというのは調べていく中で発想を得た。消すのが無理だったので逆に増やすという発想。  
- そこからもう一個のリストを消すべき要素として長さを引くというロジックを思いつけた。  
- 別解  

```python
N = int(input())
S = input()

ans = 0
for i in range(N):#N回ループ
    if S[i] == "x":
        # 左に o があるかチェック（自分が左端でない場合）
        has_left_o = (i > 0 and S[i - 1] == "o")
        
        # 右に o があるかチェック（自分が右端でない場合）
        has_right_o = (i < N - 1 and S[i + 1] == "o")
        
        # どちらにも o がない場合（両方oでない）をカウント
        if not (has_left_o or has_right_o):
            ans += 1

print(ans)
```

## B - Grandma's Footsteps 【2026/8/3 ADT EASY2】
🟢 **Status: 自力AC**(20分)

```python
S, A, B, X = map(int, input().split())
ans = 0
n = 0
count_A = 0

while X != 0:
  if X >= A:
    X -= A
    count_A += 1
  else:
    n = X
    break
  if X >= B:
    X -= B
  else:
    break

ans = (S * count_A * A) + (S * n) 

print(ans) 
```
